```python
import pandas as pd
s = pd.read_csv('../results/tables/method_benchmark_summary.csv')
s.head(10)[['famille', 'methode', 'KGE_cal', 'KGE_val', 'surajust']]
```

`surajust` is the calibration-to-test loss, and it orders the field
better than the fit does: the configurations that lose the most are the
ones that fitted the calibration window best. The convex schemes lose
about a hundredth; gradient boosting, first on calibration, loses 0.14
and finishes twenty-seventh.
